## What is Jupyter? What is Colab?

**Jupyter** is an open-source environment for writing executable documents that mix Python code, formatted prose, equations, and visualisations. A Jupyter document — a *notebook*, with the `.ipynb` extension — is a sequence of *cells* that you run interactively in a browser-based editor (JupyterLab or the classic Jupyter Notebook).

**Google Colab** is a hosted version of Jupyter, running for free in Google's cloud. You open notebooks straight from GitHub (using the "Open in Colab" badge at the top of this page), Google provides the Python environment, and you can switch on a free GPU or TPU when you need one. Nothing is installed on your own machine.

> **Tip.** If you have never installed Python before, start in Colab — it removes every installation step.

## Notebook anatomy

A notebook has three things you need to recognise:

- **Cells.** Each block in a notebook is either a *code cell* (Python source that you can execute) or a *Markdown cell* (formatted text). You can tell them apart by the toolbar at the top of the cell or by the dropdown that shows "Code" or "Markdown".
- **The kernel.** A single Python process runs in the background and executes every code cell. The kernel keeps state — variables, imports, function definitions — between cells, but it is independent of the cells themselves. Restarting the kernel wipes all in-memory state; the notebook source on disk is not affected.
- **Edit mode vs command mode.** Click *inside* a cell to enter edit mode (a green border appears, your keystrokes go into the cell). Press `Esc` to leave edit mode and enter command mode (a blue border, keystrokes are now keyboard shortcuts such as `A` to insert a cell above or `DD` to delete the current cell).

## Running cells

Three keyboard shortcuts cover almost every cell-execution need:

| Shortcut | What it does |
|----------|--------------|
| `Shift + Enter` | Run the current cell and move to the next one (creating one if needed). |
| `Ctrl + Enter` | Run the current cell and stay on it. |
| `Alt + Enter` | Run the current cell and insert a new empty cell below. |

> **Note.** The square brackets next to a code cell show its execution status: `[ ]` not yet run, `[*]` currently running, `[N]` the N-th cell executed in this kernel session. If you see `[5]` followed by `[2]`, the cells were not run top-to-bottom.

## Markdown cells with a live example

### Headings, bold, lists, and math

A numbered list:

## Magic and shell commands

IPython (the kernel that powers Jupyter) recognises three special prefixes:

- `%magic` — a *line magic* that applies to a single line, e.g. `%timeit sum(range(100))` measures execution time.
- `%%magic` — a *cell magic* that applies to the whole cell when placed on the first line, e.g. `%%time` reports how long the cell took.
- `!command` — runs the rest of the line in the underlying shell, e.g. `!pip list`.

Illustrative lines you could paste into a code cell:

```python
%timeit sum(range(1_000))
!echo "this runs in the shell"
```

> **Note.** Magics are an IPython feature, so they work identically in JupyterLab, the classic Jupyter Notebook, and Colab — but not in plain Python scripts.

## Reproducibility tips

Three habits keep your notebooks reliable:

- **Run top-to-bottom.** Do not depend on the order in which you happened to execute cells while exploring. Test with **Restart Kernel and Run All**.
- **Seed every stochastic library.** Set `np.random.seed(SEED)`, `random.seed(SEED)`, `torch.manual_seed(SEED)`, and any framework-specific seed before you train or sample.
- **Pin your dependencies.** Use a `requirements.txt` with version pins (`pandas==2.2.2`, not `pandas`) so the notebook still runs months from now.

## Summary

- Jupyter is the open-source notebook environment; Colab is Google's free hosted version. Both run the same `.ipynb` files.
- A notebook is a sequence of code and Markdown cells that share a single Python kernel.
- Three shortcuts cover most of what you need: `Shift + Enter`, `Ctrl + Enter`, `Alt + Enter`.
- Code cells produce text, rich HTML (e.g. `DataFrame.head()`), or inline figures.
- IPython adds `%magic`, `%%magic`, and `!shell` prefixes for extra functionality.
- For reproducible work: restart-and-run-all, seed your randomness, pin your dependencies.
